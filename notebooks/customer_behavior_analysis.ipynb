# Customer Shopping Behavior Analysis

Reproducible EDA and preprocessing workflow for the Customer Shopping Behavior dataset.

**Pipeline:** CSV → validation → cleaning → feature engineering → PostgreSQL → SQL analytics → Power BI

```python
from pathlib import Path
import pandas as pd

from customer_analytics.transform import clean_customer_data

DATA_PATH = Path('../data/customer.csv')
df_raw = pd.read_csv(DATA_PATH)
df_raw.shape
```

## 1. Initial data quality review

```python
display(df_raw.head())
df_raw.info()
display(df_raw.describe(include='all').T)
display(df_raw.isna().sum().sort_values(ascending=False))
```

## 2. Cleaning and feature engineering

The reusable transformation module standardizes column names, imputes missing review ratings using category medians, creates quartile-based age groups, maps purchase frequency to days, and removes the redundant promo-code field after consistency validation.

```python
df = clean_customer_data(df_raw)
display(df.head())
print(df.shape)
print('Missing values after cleaning:', int(df.isna().sum().sum()))
```

## 3. Business-oriented EDA

```python
display(df.groupby('category')['purchase_amount'].agg(['count', 'sum', 'mean']).sort_values('sum', ascending=False))
display(df.groupby('subscription_status')['purchase_amount'].agg(['count', 'mean', 'sum']))
display(df.groupby('age_group')['purchase_amount'].sum().sort_values(ascending=False))
```
